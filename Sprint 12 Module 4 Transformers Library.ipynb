## Transformers Library

```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")

result = classifier("I love learning AI!")

print(result)
```

Why is Transformers used?

Transformers is used to easily load, use, fine-tune, and train pre-trained AI models for different tasks.

| Why?                          | Purpose                                                                   |
| ----------------------------- | ------------------------------------------------------------------------- |
| **1. Use pre-trained models** | We can use models that are already trained instead of training from zero. |
| **2. Easy inference**         | Make predictions with very little code.                                   |
| **3. Fine-tuning**            | Adapt a pre-trained model to our specific dataset/task.                   |
| **4. Many AI tasks**          | Supports text, image, audio, and multimodal tasks.                        |
| **5. Saves time**             | Avoids building and training models from scratch.                         |
| **6. Supports large models**  | Provides tools to work with modern Transformer-based models.              |
| **7. Easy model loading**     | Models can be loaded directly from Hugging Face Hub.                      |


```python
Model Classes in Hugging Face Transformers

Model classes are Python classes provided by the Transformers library that let us load and work with specific types of pre-trained models.

Simple idea:

Model class = Python interface used to load a particular type of AI model.

Common Model Classes
```

| Model Class                          | Used for                       | Example                  |
| ------------------------------------ | ------------------------------ | ------------------------ |
| `AutoModel`                          | Base/general Transformer model | BERT                     |
| `AutoModelForSequenceClassification` | Text classification            | Sentiment analysis       |
| `AutoModelForTokenClassification`    | Token-level classification     | NER                      |
| `AutoModelForQuestionAnswering`      | Question answering             | Extract answer from text |
| `AutoModelForCausalLM`               | Text generation                | GPT/Llama-style models   |
| `AutoModelForSeq2SeqLM`              | Text-to-text generation        | T5, BART                 |
| `AutoModelForSpeechSeq2Seq`          | Speech-to-text                 | Whisper                  |
| `AutoModelForImageClassification`    | Image classification           | ViT                      |
