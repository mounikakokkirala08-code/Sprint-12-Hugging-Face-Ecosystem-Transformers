```python
1.Loading Pre-trained Weights

Definition: Pre-trained weights are parameters a model learned during training on large amounts of data.

from transformers import AutoModel

model = AutoModel.from_pretrained("bert-base-uncased")

Explanation: from_pretrained() loads the model architecture and its learned weights from the Hugging Face model repository.

2. Model Configuration

Definition: Model configuration contains information about the model's architecture and settings, such as hidden size, number of layers, and attention heads.

from transformers import AutoConfig

config = AutoConfig.from_pretrained("bert-base-uncased")

print("Model type:", config.model_type)
print("Hidden size:", config.hidden_size)
print("Number of layers:", config.num_hidden_layers)
```

Concept

	

Definition

	

Example / How to analyze




Model Configuration

	

Contains the model's architecture settings and hyperparameters.

	

config.hidden_size → 768 for BERT Base.




Model Architecture

	

Defines how the model is structured, including its layers and attention mechanisms.

	

BertModel uses Transformer encoder layers.




Number of Parameters

	

Total number of trainable and non-trainable numerical values in the model.

	

sum(p.numel() for p in model.parameters()) counts parameters.




Input Requirements

	

Specifies the format of data the model expects.

	

BERT expects token IDs and can use an attention mask.




Output Structure

	

Defines the format and shape of the model's output.

	

outputs.last_hidden_state contains a representation for each input token.




Print Model Output

	

Displays the model's output tensors or their shapes.

	

print(outputs.last_hidden_state.shape) shows batch size, sequence length, and hidden size.




Analyze Model Output

	

Explains what the output values represent, not just their shape.

	

Each token gets a 768-dimensional contextual representation in BERT Base.
