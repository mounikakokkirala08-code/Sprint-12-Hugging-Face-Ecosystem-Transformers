```python
What are Logits?
Definition: Logits are the raw scores produced by a machine learning model before applying Softmax or Sigmoid to convert them into probabilities.
Simple words: Logits tell us how strongly a model supports each possible answer before converting the scores into probabilities.
```

```python
import torch

# Raw scores from the model
logits = torch.tensor([4.5, 1.2, 0.8])

# Convert logits into probabilities
probabilities = torch.softmax(logits, dim=0)

print("Logits:", logits)
print("Probabilities:", probabilities)

# Find the predicted class
predicted_class = torch.argmax(logits)
print("Predicted class index:", predicted_class.item())
```

```python
Why do models produce logits?
Simple answer: Models produce logits because they need to assign a score to each possible class or token before deciding which one is most likely.
Example
Suppose a model predicts the next word after “I love”.
```

```python
Softmax
Definition: Softmax is a mathematical function that converts logits (raw model scores) into probabilities between 0 and 1.
Simple words: Softmax helps a model understand which option is most likely by converting raw scores into probabilities.
```

## Demonstration: Model Prediction Flow

| Step | Process         | Result                                                    |
| ---- | --------------- | --------------------------------------------------------- |
| 1    | Model           | Receives an input image.                                  |
| 2    | Logits          | Apple = 4.5, Banana = 1.2, Orange = 0.8                   |
| 3    | Softmax         | Converts raw scores into probabilities.                   |
| 4    | Probabilities   | Apple = 94.2%, Banana = 3.5%, Orange = 2.3%               |
| 5    | Predicted Class | Apple is selected because it has the highest probability. |

Flow: Model → Logits → Softmax → Probabilities → Predicted Class

## Why is the highest probability not always correct?

| Reason                     | Explanation                                                         |
| -------------------------- | ------------------------------------------------------------------- |
| Incorrect learned patterns | The model may learn misleading features from training data.         |
| Unfamiliar data            | The input may differ from the training examples.                    |
| Similar-looking classes    | Different objects may have similar features.                        |
| Poor image quality         | Noise, lighting, or blur may confuse the model.                     |
| Overconfidence             | The model may assign a high probability to an incorrect prediction. |

Conclusion: The highest probability represents the model's strongest prediction, not guaranteed truth. We must evaluate predictions using actual labels and performance metrics such as accuracy, precision, recall, and F1-score.
