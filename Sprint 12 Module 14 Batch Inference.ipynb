## Single Inference

Definition: Single inference means processing one input at a time through a model to generate one output.

### Demonstration



### Flow

Input 1

Model

Output 1

Input 2

Model

Output 2

Input 3

Model

Output 3

### Python Example

```
from transformers import pipelineclassifier = pipeline("sentiment-analysis")# Single inference: one input at a timeresult1 = classifier("I love Python")result2 = classifier("This is bad")result3 = classifier("Amazing experience")print(result1)print(result2)print(result3)
```

### Key Points

| Advantage                                | Limitation                              |
| ---------------------------------------- | --------------------------------------- |
| Simple to implement                      | Repeated calls may add overhead         |
| Easy to debug                            | Can be inefficient for many inputs      |
| Useful for real-time individual requests | May have lower throughput than batching |

Remember: Single inference processes one input per call. When multiple inputs are sent together in one model call, it is called batch inference.

| Input                         | Model | Output             |
| ----------------------------- | ----- | ------------------ |
| Input 1: "I love Python"      | Model | Output 1: POSITIVE |
| Input 2: "This is bad"        | Model | Output 2: NEGATIVE |
| Input 3: "Amazing experience" | Model | Output 3: POSITIVE |

````python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")

# Single inference: one input at a time
result1 = classifier("I love Python")
result2 = classifier("This is bad")
result3 = classifier("Amazing experience")

print(result1)
print(result2)
print(result3)
````
