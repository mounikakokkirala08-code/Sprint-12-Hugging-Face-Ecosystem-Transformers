## Task-Specific Auto Classes

| Auto Class                           | Definition                                                                                   | Use Case                                                         | Example                                 |
| ------------------------------------ | -------------------------------------------------------------------------------------------- | ---------------------------------------------------------------- | --------------------------------------- |
| `AutoModel`                          | Loads a pretrained model that produces hidden representations.                               | Extracting text features and embeddings.                         | Represent a sentence numerically.       |
| `AutoModelForSequenceClassification` | Loads a model designed to classify an entire text sequence.                                  | Sentiment analysis, spam detection.                              | Predict positive or negative sentiment. |
| `AutoModelForTokenClassification`    | Loads a model designed to classify each token separately.                                    | Named Entity Recognition (NER), identifying names and locations. | Identify “Ravi” as a person.            |
| `AutoModelForCausalLM`               | Loads a model that predicts the next token based on previous tokens.                         | Text generation and chatbots.                                    | Generate the next token in a sentence.  |
| `AutoModelForSeq2SeqLM`              | Loads a sequence-to-sequence model that generates an output sequence from an input sequence. | Translation, summarization.                                      | Convert English text into French.       |


```python
from transformers import AutoTokenizer, AutoModel
import torch

name = "bert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(name)
model = AutoModel.from_pretrained(name)

inputs = tokenizer("I love AI", return_tensors="pt")

model.eval()
with torch.no_grad():
    outputs = model(**inputs)

print("Output shape:", outputs.last_hidden_state.shape)
print("First token representation:",
      outputs.last_hidden_state[0, 0, :5])
```
