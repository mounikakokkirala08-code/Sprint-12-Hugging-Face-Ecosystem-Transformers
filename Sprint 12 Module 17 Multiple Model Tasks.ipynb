```python
from transformers import pipeline

# Load a pretrained text-generation model
generator = pipeline(
    "text-generation",
    model="distilgpt2"
)

# Give a prompt
prompt = "Artificial Intelligence is useful because"

# Generate text
result = generator(
    prompt,
    max_new_tokens=40,
    do_sample=False
)

print(result[0]["generated_text"])
```

## Step 4: Question Answering

```python
from transformers import pipeline

# Load a pretrained question-answering model
qa = pipeline(
    "question-answering",
    model="distilbert-base-cased-distilled-squad"
)

# Provide context
context = """
Python is a programming language.
It is widely used in data science, machine learning,
automation, and web development.
"""

# Ask a question
question = "What is Python widely used for?"

# Get the answer
result = qa(
    question=question,
    context=context
)

print("Answer:", result["answer"])
print("Confidence:", result["score"])
```

```python
from transformers import pipeline

# Load a pretrained NER model
ner = pipeline(
    "ner",
    aggregation_strategy="simple"
)

# Input sentence
text = "Sundar Pichai works at Google in California."

# Identify entities
results = ner(text)

# Display the results
for entity in results:
    print(
        "Text:", entity["word"],
        "| Entity:", entity["entity_group"],
        "| Score:", round(entity["score"], 3)
    )
```

Hugging Face Model Comparison — Practical Documentation
Objective: Measure and compare two pretrained Hugging Face models performing the same AI task, using model size, loading time, inference time, memory consumption, and batch performance.
We will compare two sentiment classification models:
- Model 1: distilbert/distilbert-base-uncased-finetuned-sst-2-english
- Model 2: google/bert_uncased_L-2_H-128_A-2
The first is a DistilBERT sentiment model. The second is a compact BERT model; we will attach a sentiment-classification head for this experiment and fine-tune it on a small sample of SST-2 data before evaluating it. To keep the comparison fair, both models should use the same dataset, hardware, batch sizes, and evaluation procedure.
For a simpler first experiment, you can compare two already fine-tuned sentiment models instead. The code below uses the first approach for a meaningful task comparison.

2. Complete benchmarking code
Run this in Google Colab or Jupyter Notebook. The models are downloaded on the first run, so loading time will depend on your internet connection and cache.

```python
import os
import time
import gc
import psutil
import torch

from huggingface_hub import snapshot_download
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)

# Two models trained for the same sentiment task
MODELS = {
    "DistilBERT": (
        "distilbert/distilbert-base-uncased-finetuned-sst-2-english"
    ),
    "BERT-small": (
        "textattack/bert-base-uncased-SST-2"
    )
}

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

# Sample inputs for testing
texts = [
    "I love this product. It works perfectly.",
    "This service is terrible and disappointing.",
    "The experience was good and enjoyable.",
    "I am unhappy with the quality.",
    "The product is excellent.",
    "The service was not good.",
    "I am very satisfied with the results.",
    "I would not recommend this product."
] * 20  # 160 sentences for batch testing


def memory_mb():
    """Return current Python process RAM in MB."""
    process = psutil.Process(os.getpid())
    return process.memory_info().rss / (1024 ** 2)


def folder_size_mb(path):
    """Calculate downloaded model file size."""
    total = 0
    for root, dirs, files in os.walk(path):
        for filename in files:
            file_path = os.path.join(root, filename)
            if os.path.isfile(file_path):
                total += os.path.getsize(file_path)
    return total / (1024 ** 2)


def benchmark_model(model_name, model_id):
    print(f"\nBenchmarking {model_name}")

    gc.collect()

    # Download model files and measure their total size
    download_path = snapshot_download(model_id)
    model_size = folder_size_mb(download_path)

    # Measure model loading time and RAM usage
    ram_before = memory_mb()
    start = time.perf_counter()

    tokenizer = AutoTokenizer.from_pretrained(model_id)
    model = AutoModelForSequenceClassification.from_pretrained(
        model_id
    )
    model.to(device)
    model.eval()

    loading_time = time.perf_counter() - start
    ram_after_load = memory_mb()
    ram_increase = ram_after_load - ram_before

    # Warm-up: avoid including first-run overhead in inference
    warmup_inputs = tokenizer(
        texts[:2],
        padding=True,
        truncation=True,
        return_tensors="pt"
    ).to(device)

    with torch.inference_mode():
        model(**warmup_inputs)

    if device.type == "cuda":
        torch.cuda.synchronize()
        torch.cuda.reset_peak_memory_stats()

    inference_results = []

    # Test batch sizes
    for batch_size in [1, 8, 16, 32]:
        total_time = 0
        processed = 0

        for start_idx in range(0, len(texts), batch_size):
            batch = texts[start_idx:start_idx + batch_size]

            inputs = tokenizer(
                batch,
                padding=True,
                truncation=True,
                return_tensors="pt"
            ).to(device)

            if device.type == "cuda":
                torch.cuda.synchronize()

            t0 = time.perf_counter()

            with torch.inference_mode():
                outputs = model(**inputs)

            if device.type == "cuda":
                torch.cuda.synchronize()

            total_time += time.perf_counter() - t0
            processed += len(batch)

        throughput = processed / total_time
        average_latency = total_time / processed * 1000

        inference_results.append({
            "batch_size": batch_size,
            "total_time_sec": round(total_time, 3),
            "avg_latency_ms_per_text": round(average_latency, 2),
            "throughput_texts_per_sec": round(throughput, 2)
        })

    # Record memory after inference
    ram_after_inference = memory_mb()

    if device.type == "cuda":
        peak_gpu_mb = (
            torch.cuda.max_memory_allocated() / (1024 ** 2)
        )
    else:
        peak_gpu_mb = None

    results = {
        "model": model_name,
        "model_size_mb": round(model_size, 2),
        "loading_time_sec": round(loading_time, 3),
        "ram_increase_mb": round(ram_increase, 2),
        "ram_after_load_mb": round(ram_after_load, 2),
        "ram_after_inference_mb": round(ram_after_inference, 2),
        "peak_gpu_memory_mb": (
            round(peak_gpu_mb, 2)
            if peak_gpu_mb is not None else "N/A (CPU)"
        ),
        "batch_results": inference_results
    }

    # Free resources before testing the next model
    del model, tokenizer
    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return results


# Run benchmarks
all_results = []

for name, model_id in MODELS.items():
    try:
        result = benchmark_model(name, model_id)
        all_results.append(result)
        print(result)
    except Exception as error:
        print(f"{name} failed: {error}")
```

```python
import pandas as pd

# Compare model size, loading time, and memory
summary = pd.DataFrame([
    {
        "Model": r["model"],
        "Model Size (MB)": r["model_size_mb"],
        "Loading Time (sec)": r["loading_time_sec"],
        "RAM Increase (MB)": r["ram_increase_mb"],
        "RAM After Load (MB)": r["ram_after_load_mb"],
        "RAM After Inference (MB)": r["ram_after_inference_mb"],
        "Peak GPU Memory (MB)": r["peak_gpu_memory_mb"]
    }
    for r in all_results
])

display(summary)

# Compare batch performance

batch_rows = []

for result in all_results:
    for batch in result["batch_results"]:
        batch_rows.append({
            "Model": result["model"],
            "Batch Size": batch["batch_size"],
            "Total Time (sec)": batch["total_time_sec"],
            "Latency (ms/text)": batch["avg_latency_ms_per_text"],
            "Throughput (texts/sec)": batch["throughput_texts_per_sec"]
        })

batch_comparison = pd.DataFrame(batch_rows)
display(batch_comparison)
```
