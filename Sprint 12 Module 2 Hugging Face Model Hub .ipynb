```python
### Model 1 — DistilBERT SST-2

Model: distilbert/distilbert-base-uncased-finetuned-sst-2-english
Author/Organization: DistilBERT / Hugging Face model community

Task

Text classification → Sentiment analysis

Example:

"I really enjoyed this movie."
             ↓
          POSITIVE
Architecture

DistilBERT, a smaller/distilled version of BERT using a Transformer encoder.

Parameters

67 million

Model size

The Hugging Face page reports 67M parameters, F32.

Language

English

Training information

The model is a fine-tuned DistilBERT checkpoint trained on the Stanford Sentiment Treebank (SST-2). The listed fine-tuning configuration includes learning rate 1e-5, batch size 32, maximum sequence length 128, and 3 epochs.

Input format

Text:

"I love this product!"
Output format
POSITIVE
0.99
License

Apache-2.0

Intended use

Sentiment classification, especially positive/negative English text.

Limitations
English only
Designed specifically for sentiment classification
May perform poorly on domains very different from its training data
Bias should be evaluated for the intended application; the model card specifically recommends probing bias.
Hardware

Because it is only 67M parameters, it is relatively lightweight and can be used for CPU inference for many applications; a GPU can improve throughput for larger workloads or batches.

Best choice when: You need fast English sentiment analysis with relatively low resource requirements.
```

```python
### Model 2 — BERT Base Uncased

Model: google-bert/bert-base-uncased
Author/Organization: Google

Task

Masked language modeling / language representation

Example:

"The capital of France is [MASK]."
                       ↓
                    Paris
Architecture

BERT Transformer encoder

The configuration contains 12 layers, 12 attention heads, and 768 hidden dimensions.

Parameters

110M

Model size

Approximately 110M parameters.

Language

English

Training information

BERT was pretrained on English using a masked language modeling objective and was trained using datasets including BookCorpus and Wikipedia.

Input format

Text containing tokens, typically processed by the BERT tokenizer:

"The capital of France is [MASK]."
Output format

Predicted token probabilities/logits for the masked position.

License

Apache-2.0

Intended use

Use as a base language representation model and fine-tune it for downstream NLP tasks such as:

Classification
NER
Question answering
Feature extraction
Limitations
English-focused
Not a conversational/generative chatbot
Base BERT needs task-specific fine-tuning for many practical applications
Has the limitations and biases present in its training data
Hardware

CPU can be used, but GPU is preferred for faster inference/fine-tuning.

Best choice when: You need a general English NLP encoder and want to fine-tune it for a specific task.
```

```python
odel 3 — BART Large MNLI

Model: facebook/bart-large-mnli
Author/Organization: Meta / Facebook AI

Task

Zero-shot classification / Natural Language Inference

Example:

Text:
"Apple released a new laptop."

Labels:
["Technology", "Sports", "Finance"]

             ↓

Technology → highest score
Architecture

BART Transformer encoder-decoder

The configuration has 12 encoder layers + 12 decoder layers, with 16 attention heads and 1024-dimensional hidden representations.

Parameters

Approximately 407M parameters, based on the BART-large architecture plus classification head.

Model size

The Hugging Face repository is approximately 1.63 GB for the listed model files.

Language

English

Training information

It is based on BART and fine-tuned on the Multi-Genre Natural Language Inference (MNLI) task.

Input format
text = "AI is changing software development."

candidate_labels = [
    "Technology",
    "Sports",
    "Business"
]
Output format
labels:
Technology
Business
Sports

scores:
0.94
0.04
0.02
License

MIT

Intended use

Zero-shot classification where you provide your own candidate labels without training a new classification head.

Limitations
Large compared with DistilBERT
More computationally expensive
Classification quality depends on how candidate labels are formulated
Not ideal when you require extremely low latency
Hardware

GPU recommended for production/high-throughput use. CPU inference is possible but can be considerably slower.

Best choice when: You need classification without having a labeled training dataset for every category.
```

```python
Model 4 — Whisper Small

Model: openai/whisper-small
Author/Organization: OpenAI

Task

Automatic Speech Recognition (ASR) and speech translation.

Architecture

Transformer encoder-decoder / sequence-to-sequence

Parameters

244M

Model size

Approximately 1 GB in the commonly used configuration.

Language

Multilingual — 98 languages represented in its training data, with Whisper supporting multilingual speech recognition/translation.

Training information

Whisper was trained using approximately 680,000 hours of labelled audio and transcripts collected from the internet.

Input format
Audio file / speech
       ↓
Whisper
Output format
"Welcome to our company."

It can also produce translated text depending on the task.

License

Apache-2.0

Intended use
Call transcription
Meeting transcription
Voice-to-text
Speech translation
Limitations
Accuracy varies by language
Performance depends on audio quality, accents, background noise, and domain-specific terminology
Larger Whisper models require significantly more compute
Training-data representation differs by language; the model card notes that recognition performance correlates with the amount of training data for a language.
Hardware

Can run on CPU, but GPU is strongly preferred for faster transcription, especially for real-time or batch workloads.

Best choice when: Your input is speech/audio and you need transcription.
```

```python
Model 5 — Vision Transformer (ViT)

Model: google/vit-base-patch16-224
Author/Organization: Google

Task

Image classification

Architecture

Vision Transformer (ViT)

Instead of processing an image exactly like a traditional CNN, ViT divides the image into patches and processes them using Transformer mechanisms.

Parameters

86.6M

Model size

Approximately 86.6M parameters.

Language

Not applicable — this is a vision model.

Training information

The model was:

Pre-trained on ImageNet-21k
14 million images
21,843 classes
Fine-tuned on ImageNet 2012
1 million images
1,000 classes

Both use 224×224 image resolution.

Input format
Image
↓
224 × 224 image
Output format

Predicted ImageNet class and scores.

Labrador retriever → 0.92
Golden retriever   → 0.04
...
License

Apache-2.0

Intended use

Image classification and transfer learning for downstream vision tasks.

Limitations
Designed for image classification
Not directly a general object-detection model
Input preprocessing/resolution matters
Performance can change significantly when the target domain differs from ImageNet
Hardware

CPU inference is possible, but GPU is preferred for high-throughput inference or fine-tuning.

Best choice when: Your input is images and the task is classification.
```
