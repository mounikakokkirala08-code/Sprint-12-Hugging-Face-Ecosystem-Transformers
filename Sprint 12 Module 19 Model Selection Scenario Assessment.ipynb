# AI Model Selection — Practical Assignment

Objective: Understand a real-world problem, identify the required AI task, compare Hugging Face models, and justify the best model for the scenario.

The key is not to choose the biggest model. Choose a model that balances accuracy, speed, hardware requirements, cost, and limitations.

## Scenario 1: Customer Support Call Transcription

Problem: A company wants to automatically convert recorded customer support calls into text.

| Requirement             | Analysis                                                                                                                   |
| ----------------------- | -------------------------------------------------------------------------------------------------------------------------- |
| Required model type     | Automatic Speech Recognition (ASR)                                                                                         |
| Input                   | Audio recordings, such as WAV or MP3                                                                                       |
| Expected output         | Text transcript of the conversation                                                                                        |
| Candidate models        | `openai/whisper-tiny.en`, `openai/whisper-small.en`                                                                        |
| Hardware requirements   | CPU can run inference, but GPU generally improves speed; memory depends on model size and audio length                     |
| Accuracy considerations | Accents, background noise, call quality, domain-specific terms, and overlapping speakers                                   |
| Limitations             | May misrecognize names, numbers, and technical terms; standard transcription does not reliably identify speakers by itself |

Model references: Whisper family on Hugging Face.

Model comparison

- `whisper-tiny.en`: smaller and faster; suitable for resource-constrained applications.
- `whisper-small.en`: larger and typically more accurate, but requires more computation.

Selection: Start with `whisper-small.en` if transcription quality is the priority and hardware supports it. Choose `whisper-tiny.en` when low latency and limited resources matter more. Evaluate both using actual call recordings.

## Scenario 2: Document Classification

Problem: A company receives thousands of documents and needs to categorize them as invoices, contracts, insurance claims, or complaints.

| Requirement             | Analysis                                                                                                                                                                         |
| ----------------------- | -------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| Required model type     | Text classification                                                                                                                                                              |
| Input                   | Document text or extracted PDF text                                                                                                                                              |
| Expected output         | A document category and confidence score                                                                                                                                         |
| Candidate models        | `distilbert/distilbert-base-uncased`, `FacebookAI/roberta-base`                                                                                                                  |
| Hardware requirements   | CPU is suitable for smaller workloads; GPU helps with high-volume processing                                                                                                     |
| Accuracy considerations | Document length, OCR quality, class imbalance, and similarity between categories                                                                                                 |
| Limitations             | Base models are not automatically trained for these business categories; they need a suitable classification head and fine-tuning or another appropriate classification approach |

Model comparison

- DistilBERT: generally lighter and faster.
- RoBERTa-base: larger; may perform better after task-specific fine-tuning, but that is not guaranteed.

Selection: Start with DistilBERT for a lightweight classification service. Fine-tune and compare both models using labeled company documents before deployment.

## Scenario 3: Customer Sentiment Analysis

Problem: A company wants to analyze customer reviews and identify positive, negative, or neutral feedback.

| Requirement             | Analysis                                                                                                                                                                       |
| ----------------------- | ------------------------------------------------------------------------------------------------------------------------------------------------------------------------------ |
| Required model type     | Sentiment classification                                                                                                                                                       |
| Input                   | Customer reviews, emails, or feedback                                                                                                                                          |
| Expected output         | Sentiment label and confidence score                                                                                                                                           |
| Candidate models        | `distilbert/distilbert-base-uncased-finetuned-sst-2-english`, `cardiffnlp/twitter-roberta-base-sentiment-latest`                                                               |
| Hardware requirements   | CPU for small workloads; GPU for high-volume, low-latency workloads                                                                                                            |
| Accuracy considerations | Sarcasm, mixed sentiment, domain-specific language, and short reviews                                                                                                          |
| Limitations             | The DistilBERT candidate predicts positive or negative sentiment, not neutral. The Twitter RoBERTa model is trained on social media text, which may not match customer reviews |

Model comparison

- DistilBERT SST-2: lightweight, fast, and suitable for binary sentiment.
- Twitter RoBERTa: supports negative, neutral, and positive labels and may better suit social-media-style text.

Selection: Choose Twitter RoBERTa if three sentiment categories are required and its performance is suitable for your data. Validate it on real customer reviews before deployment.

## Scenario 4: Entity Extraction

Problem: An insurance company wants to extract customer names, organization names, and locations from claim documents.

| Requirement             | Analysis                                                                                                          |
| ----------------------- | ----------------------------------------------------------------------------------------------------------------- |
| Required model type     | Named Entity Recognition (NER)                                                                                    |
| Input                   | Claim descriptions and document text                                                                              |
| Expected output         | Entities with their types and text positions                                                                      |
| Candidate models        | `dslim/bert-base-NER`, `dbmdz/bert-large-cased-finetuned-conll03-english`                                         |
| Hardware requirements   | CPU for smaller workloads; GPU for high-volume processing                                                         |
| Accuracy considerations | Names, abbreviations, spelling errors, OCR errors, and insurance-specific terminology                             |
| Limitations             | General NER models may not recognize policy numbers, claim IDs, or domain-specific entities without customization |

Model comparison

- BERT-base NER: moderate resource requirements and a practical starting point.
- BERT-large NER: larger and more computationally expensive; may perform differently depending on the data.

Selection: Start with BERT-base NER for general people, organizations, and locations. Fine-tune it on annotated insurance documents if custom entity types are needed.

## Scenario 5: Text Summarization

Problem: A company wants to summarize lengthy insurance reports so employees can review key points quickly.

| Requirement             | Analysis                                                                                                                |
| ----------------------- | ----------------------------------------------------------------------------------------------------------------------- |
| Required model type     | Abstractive text summarization                                                                                          |
| Input                   | Long reports, claim descriptions, or business documents                                                                 |
| Expected output         | A shorter summary containing the main points                                                                            |
| Candidate models        | `sshleifer/distilbart-cnn-12-6`, `facebook/bart-large-cnn`                                                              |
| Hardware requirements   | CPU for low-volume use, although it may be slow; GPU recommended for larger models and high throughput                  |
| Accuracy considerations | Factual consistency, important details, numbers, names, and coverage of key points                                      |
| Limitations             | Models can omit important details or generate unsupported statements. Input length limits may require document chunking |

Model comparison

- DistilBART: smaller and generally faster, making it useful for resource-constrained environments.
- BART-large-CNN: larger and potentially better at some summarization tasks, but actual quality depends on the document type.

Selection: Start with DistilBART for efficient summarization. Compare it with BART-large-CNN on real reports and verify that summaries preserve critical facts.
