# Sprint 12 – CPU vs GPU Inference

## 1. Understand the Concepts

| Topic                 | Simple Explanation                                                                                               |
| --------------------- | ---------------------------------------------------------------------------------------------------------------- |
| CPU inference         | Runs model predictions using the computer's CPU. It works on most computers but may be slower for large models.  |
| GPU inference         | Uses a Graphics Processing Unit to process many operations in parallel. It can speed up deep learning inference. |
| Device selection      | Choosing whether to run the model on the CPU or GPU.                                                             |
| Tensor device         | The device where input data tensors are stored.                                                                  |
| Model device          | The device where the model's parameters are stored.                                                              |
| Memory considerations | CPU uses system RAM; GPU uses GPU memory (VRAM). Both have limited capacity.                                     |

## 4. Device Selection and Memory

| Factor                | CPU                                      | GPU                                          |
| --------------------- | ---------------------------------------- | -------------------------------------------- |
| Hardware              | General-purpose processor                | Graphics processor                           |
| Memory                | System RAM                               | GPU VRAM                                     |
| Large model inference | Can be slower                            | Often faster if sufficient VRAM is available |
| Availability          | Available on most computers              | Requires a compatible GPU and software       |
| Best use              | Small models, testing, general inference | Deep learning and high-throughput inference  |

Important: The model and input tensors must be on compatible devices. If the model is on a GPU but the input tensors are on the CPU, PyTorch commonly raises a device mismatch error.

## 5. Check Memory Usage

```

Running on CPU.")
```

- Allocated memory: GPU memory currently used by PyTorch tensors.
- Reserved memory: GPU memory held by PyTorch's caching allocator, including unused cached blocks.

## 6. Key Takeaways

1. CPU inference is convenient and widely available.
2. GPU inference can improve speed for computationally intensive models.
3. The model and its input tensors must be on the correct device.
4. Monitor RAM and VRAM to avoid memory errors.
5. Benchmark both devices when possible; GPU performance depends on the model, batch size, and hardware.

````python
import torch

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print(
        "Allocated GPU memory:",
        round(torch.cuda.memory_allocated() / 1024**2, 2),
        "MB"
    )
    print(
        "Reserved GPU memory:",
        round(torch.cuda.memory_reserved() / 1024**2, 2),
        "MB"
    )
else:
    print("GPU unavailable. Running on CPU.")
````

````python
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification

model_name = "distilbert/distilbert-base-uncased-finetuned-sst-2-english"

# Step 1: Select device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Selected device:", device)

# Step 2: Load tokenizer and model
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

# Step 3: Move model to selected device
model = model.to(device)
model.eval()

# Step 4: Prepare input
text = "I love learning Python"
inputs = tokenizer(text, return_tensors="pt")

# Step 5: Move input tensors to the same device
inputs = {key: value.to(device) for key, value in inputs.items()}

# Step 6: Run inference
with torch.no_grad():
    outputs = model(**inputs)

predicted_id = outputs.logits.argmax(dim=1).item()
prediction = model.config.id2label[predicted_id]

print("Model device:", next(model.parameters()).device)
print("Input tensor device:", inputs["input_ids"].device)
print("Prediction:", prediction)
````
