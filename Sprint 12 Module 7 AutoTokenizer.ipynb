### AutoTokenizer

What is AutoTokenizer?

Definition: AutoTokenizer automatically loads the correct tokenizer for a pretrained model from Hugging Face.

Example code:

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

text = "I love AI"

tokens = tokenizer.tokenize(text)

print(tokens)
```

```python
AutoTokenizer — loads the appropriate tokenizer.

from_pretrained() — loads a tokenizer that matches the specified pretrained model.

tokenize() — splits the text into tokens.
```

```python
Loading a Tokenizer from a Model Repository.

Definition: It means downloading and loading a pretrained model's tokenizer from the Hugging Face Hub using its model name.
```

```python
from transformers import AutoTokenizer

tokenize = AutoTokenizer.from_pretrained("bert-base-uncased")

tex = "i love you"

result = tokenize.tokenize(tex)

print(result)
```

```python
Encoding Text

Definition: Encoding converts text into numerical token IDs that a machine learning model can understand.

Example code
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

text = "I love AI"

encoded = tokenizer.encode(text)

print(encoded)
```

```python
Decoding Tokens

Definition: Decoding converts token IDs back into readable text.

Example code
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

token_ids = [101, 1045, 2293, 9932, 102]

text = tokenizer.decode(token_ids)

print(text)
```

```python
Batch Encoding

Definition: Batch encoding means encoding multiple sentences at the same time into token IDs.

Example code
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

texts = [
    "I love AI",
    "I am learning Python",
    "Generative AI is useful"
]

encoded = tokenizer(
    texts,
    padding=True,
    truncation=True
)

print(encoded["input_ids"])
print(encoded["attention_mask"])
```

```python
1. Padding

Definition: Padding adds special [PAD] tokens to shorter sequences so all sequences in a batch have the same length.

result = tokenizer(
    ["I love AI", "I am learning Generative AI"],
    padding=True
)

print(result["input_ids"])

Why? Models process batches more easily when sequences have equal lengths.
```

```python
2. Truncation

Definition: Truncation removes extra tokens when a sequence exceeds the maximum allowed length.

result = tokenizer(
    "I am learning Generative AI and machine learning",
    truncation=True,
    max_length=6
)

print(result["input_ids"])

Why? It keeps the input within the model's token-length limit.
```

```python
Attention Mask

Definition: An attention mask tells the model which tokens to process as valid input and which positions are padding.

result = tokenizer(
    ["I love AI", "I am learning Generative AI"],
    padding=True,
    return_tensors="pt"
)

print(result["attention_mask"])

Example output:
```

| Concept        | Definition                                  | Purpose                                               | Example                         |
| -------------- | ------------------------------------------- | ----------------------------------------------------- | ------------------------------- |
| Padding        | Adds `[PAD]` tokens to shorter sequences.   | Makes all sequences equal in length.                  | `padding=True`                  |
| Truncation     | Removes extra tokens from longer sequences. | Limits input to the maximum sequence length.          | `truncation=True`               |
| Attention Mask | Identifies real tokens and padding tokens.  | Helps the model distinguish valid input from padding. | `1` = real token, `0` = padding |
