## 1. What is a Token?

A token is a small unit of text that an AI model processes as input or generates as output.

A token can be:

A complete word

Part of a word

A punctuation mark

A number or symbol

Example:

Input: I love AI.

A tokenizer might split it into:

["I", "love", "AI", "."]

Here, there are 4 tokens. The exact number depends on the tokenizer and model.

2. Why do we need tokens?

Computers don't directly understand text the way humans do. Language models convert text into numerical representations that they can process.

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)

text = "I love learning AI"

tokens = tokenizer.tokenize(text)
token_ids = tokenizer.encode(text)

print("Tokens:", tokens)
print("Token IDs:", token_ids)
```

### 1. What is a Tokenizer?

A tokenizer is a component that converts text into smaller units called tokens and converts those tokens into numerical IDs that a language model can process.

Simple example:

Input text:

I love Generative AI

After tokenization, a tokenizer might produce:

["I", "love", "Generative", "AI"]

After converting to token IDs (illustrative):

[101, 1045, 2293, 102]

Actual tokens and IDs depend on the tokenizer. Some tokenizers also add special tokens automatically.

2. Why do we need a Tokenizer?

A language model works with numerical representations rather than directly processing raw text.

### 1. What is Vocabulary in Tokenization?

Vocabulary is the complete collection of tokens that a tokenizer knows and can map to numerical IDs.

Think of it like a dictionary. Every token in the vocabulary has a unique ID.

For example, a simplified vocabulary might look like this:

T

These IDs are illustrative, not actual IDs from a pretrained model.

```python
Attention Mask

Definition: An attention mask tells the model which tokens to pay attention to and which tokens to ignore, typically padding tokens.

Example:

Input IDs:      [101, 200, 300,   0,   0]
Attention Mask: [  1,   1,   1,   0,   0]

1 → Real token

0 → Padding token to ignore
```

```python
Padding

Definition: Padding adds special padding tokens to shorter sequences so that sequences in a batch have the same length.

Example:

Sentence 1: I love AI
Sentence 2: I love Generative AI

After padding:
Sentence 1: I love AI [PAD] [PAD]
Sentence 2: I love Generative AI
```

```python
Truncation

Definition: Truncation removes tokens from a sequence that exceeds the specified maximum length.

Example:

Original: I love learning Generative AI today

Max length: 4 tokens

After truncation:
I love learning Generative

Illustrative example; actual token counts depend on the tokenizer and whether special tokens are included.
```

```python
. Maximum Sequence Length

Definition: Maximum sequence length is the maximum number of tokens allowed in a model input for a particular configuration.

Example:

Maximum length = 5

Input: I love learning Generative AI today

After truncation:
I love learning Generative AI
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)

texts = [
    "I love AI",
    "I love learning Generative AI today"
]

result = tokenizer(
    texts,
    padding=True,
    truncation=True,
    max_length=6,
    return_tensors="pt"
)

print("Input IDs:")
print(result["input_ids"])

print("Attention Mask:")
print(result["attention_mask"])

print("Decoded text:")
for ids in result["input_ids"]:
    print(tokenizer.decode(ids))
```
