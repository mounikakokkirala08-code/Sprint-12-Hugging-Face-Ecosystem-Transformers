## Model Repository Structure

Model Repository Structure

A model repository structure is the collection of files and folders that are stored in a model's repository. These files contain the model weights, configuration, tokenizer, documentation, and other information required to use the model.

```python
Model Repository
│
├── config.json
├── model.safetensors
├── tokenizer.json
├── tokenizer_config.json
├── special_tokens_map.json
├── vocab.txt
├── README.md
└── LICENSE
```

| File                      | Purpose                                          |
| ------------------------- | ------------------------------------------------ |
| `config.json`             | Model architecture and settings                  |
| `model.safetensors`       | Learned model weights                            |
| `tokenizer.json`          | Tokenizer configuration/vocabulary               |
| `tokenizer_config.json`   | Tokenizer settings                               |
| `special_tokens_map.json` | Special tokens such as `[CLS]`, `[SEP]`, `[PAD]` |
| `vocab.txt`               | Vocabulary used by the tokenizer                 |
| `README.md`               | Model Card/documentation                         |
| `LICENSE`                 | Rules for using the model                        |


| Component / File                                                  | What is it?                  | Purpose                                                                                                              | Simple Example                                           |
| ----------------------------------------------------------------- | ---------------------------- | -------------------------------------------------------------------------------------------------------------------- | -------------------------------------------------------- |
| **`config.json`**                                                 | Model configuration          | Defines the **structure and settings** of the model. It tells the library how to create/load the model architecture. | Number of layers, hidden size, attention heads           |
| **Model Weight Files** (`model.safetensors`, `pytorch_model.bin`) | Learned numerical values     | Contains the **knowledge learned during training**. These weights are used to make predictions.                      | Learned patterns for recognizing sentiment               |
| **Tokenizer Files** (`tokenizer.json`, `tokenizer_config.json`)   | Tokenization information     | Defines **how input text is converted into tokens and token IDs** that the model can process.                        | `"I love AI"` → `["I", "love", "AI"]` → IDs              |
| **Vocabulary Files** (`vocab.txt`, etc.)                          | List/mapping of known tokens | Contains the **vocabulary that the tokenizer uses** to map tokens to numerical IDs.                                  | `"hello"` → `7592`                                       |
| **README / Model Card** (`README.md`)                             | Model documentation          | Explains **what the model is, how it was trained, how to use it, its limitations, intended use, and license**.       | "This model is designed for English sentiment analysis." |


```python
                 MODEL REPOSITORY
                       │
       ┌───────────────┼────────────────┐
       ↓               ↓                ↓
 config.json      Tokenizer        Model Weights
       │               │                │
       ↓               ↓                ↓
 Model structure   Input text       Learned knowledge
                       │                │
                 Vocabulary             │
                       │                │
                       └───────┬────────┘
                               ↓
                         AI Model
                               ↓
                           Output
                       
                  README / Model Card
                         ↓
                Explains everything
```

```python
Config → How the model is built
Weights → What the model learned
Tokenizer → How text is converted
Vocabulary → Which tokens are known
Model Card → How/why the model should be used
```
